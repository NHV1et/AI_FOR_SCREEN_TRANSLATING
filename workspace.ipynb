```python
import os
import zipfile

def prepare_manga_data_zip(output_zip_path):
    img_root = r"D:\AI_FOR_SCREEN_TRANSLATING\dataset\Manga109\Manga109_released_2023_12_07\images"
    lbl_root = r"D:\AI_FOR_SCREEN_TRANSLATING\dataset\List_Of_Labels"
    
    books = [d for d in os.listdir(img_root) if os.path.isdir(os.path.join(img_root, d))]
    
    with zipfile.ZipFile(output_zip_path, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for book in books:
            book_img_dir = os.path.join(img_root, book)
            # Dùng thủ thuật tìm thư mục nhãn không phân biệt hoa thường
            all_label_dirs = os.listdir(lbl_root)
            matched_lbl_dir = next((d for d in all_label_dirs if d.lower() == book.lower()), None)

            if not matched_lbl_dir:
                print(f"⚠️ CẢNH BÁO: Không tìm thấy thư mục nhãn cho bộ: {book}")
                continue
            
            book_lbl_dir = os.path.join(lbl_root, matched_lbl_dir)
            print(f"✅ Đang đóng gói: {book} (Đã khớp nhãn từ: {matched_lbl_dir})")

            # Nén Images
            for img_file in os.listdir(book_img_dir):
                if img_file.lower().endswith(('.jpg', '.png', '.jpeg')):
                    zipf.write(os.path.join(book_img_dir, img_file), 
                               os.path.join(book, "images", img_file))
            
            # Nén Labels
            for lbl_file in os.listdir(book_lbl_dir):
                if lbl_file.endswith('.txt'):
                    zipf.write(os.path.join(book_lbl_dir, lbl_file), 
                               os.path.join(book, "labels", lbl_file))

    print(f"\n🚀 Xong! File lưu tại: {output_zip_path}")

prepare_manga_data_zip(r"D:\manga_data_for_train_and_val.zip")
```
